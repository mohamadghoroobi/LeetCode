## Problem: Best Time to Buy and Sell Stock

You are given an array `prices` where `prices[i]` is the price of a given stock on the `i-th` day.

You want to maximize your profit by choosing a **single day** to buy one stock and choosing a **different day in the future** to sell that stock.

Return the maximum profit you can achieve from this transaction. If you cannot achieve any profit, return `0`.

### Example 1:
```
Input: prices = [7,1,5,3,6,4]
Output: 5
Explanation: Buy on day 2 (price = 1) and sell on day 5 (price = 6), profit = 6-1 = 5.
Note that buying on day 2 and selling on day 1 is not allowed because you must buy before you sell.
```

### Example 2:
```
Input: prices = [7,6,4,3,1]
Output: 0
Explanation: In this case, no transactions are done and the max profit = 0.
```

### Constraints:
- `1 <= prices.length <= 10^5`
- `0 <= prices[i] <= 10^4`

```python
def max_profit(prices):
    pass
```

## 1st Try

````python
def max_profit(prices):
    best_profit = 0
    left = 0
    right = len(prices) - 1

    while left < right:
        current_profit = prices[right] - prices[left]

        if current_profit > best_profit:
            best_profit = current_profit

        if prices[right - 1] >= prices[right]:
            right -= 1

        else:
            left += 1

    return best_profit
````

````python
# Input: prices = [7,1,5,3,6,4]
# Output: 5

prices = [7, 1, 5, 3, 6, 4]
max_profit(prices)
````

````python
# Input: prices = [7,6,4,3,1]
# Output: 0

prices = [7, 6, 4, 3, 1]
max_profit(prices)
````

### ❌ Bug: Logic is flawed
Your two-pointer approach doesn't work for this problem. You're moving `left` and `right` based on comparing adjacent prices, which misses the core idea.

### 🔍 Test case that fails:
```

Output: 0
Expected: 2 (buy at 2, sell at 4)
```

### 📝 Why it fails:
- You start with `left=0, right=2` → profit = `1-2 = -1`
- `prices[1] (4) >= prices[2] (1)` → `right` moves to 1
- `left=0, right=1` → profit = `4-2 = 2` → `best_profit = 2`
- But wait, that actually works for this one. Try this:

### 💡 Hint:
This isn't a two-pointer problem. Think simpler:
- You want the **minimum price seen so far** as the buy day
- At each step, calculate profit if you sell **today** at current price
- Track the maximum profit

## 2nd Try

````python
def max_profit_v2(prices):
    # sell = len(prices) - 1
    last_day_buy = len(prices) - 2
    best_buy = min(prices[:last_day_buy])
    best_buy_day = prices.index(best_buy)
    best_sell = max(prices[best_buy_day + 1:])
    return max(best_sell - best_buy, 0)
````

## Review: Best Time to Buy and Sell Stock v2

### ❌ Bug: Logic is incorrect
You're finding the global minimum and global maximum after that point, but the global minimum might not be the optimal buy day.

### 🔍 Test case that fails:
```
prices = [2, 10, 1, 5]
Your output: 4 (buy at 1, sell at 5)
Expected: 8 (buy at 2, sell at 10)
```
You missed the bigger profit earlier because you locked into the global minimum.

### ❌ Bug: Index slicing issue
`prices[:last_day_buy]` — you force buying before the second-to-last day, which is unnecessary. You can buy on the last day (and just get 0 profit).

### 💡 Hint:
Don't find the minimum of the whole array. You need to track the **minimum price seen so far** as you move left to right. At each step:
1. Update `min_price = min(min_price, current_price)`
2. Calculate `profit = current_price - min_price`
3. Update `max_profit = max(max_profit, profit)`

This is a **single pass**, one pointer, no slicing. Try again.

## 3rd Try

````python
def max_profit_v3(prices):
    min_price = 1000
    max_profit = 0
    for current_price in prices:
        min_price = min(min_price, current_price)
        profit = current_price - min_price
        max_profit = max(max_profit, profit)
    return max_profit
````

## Review: Best Time to Buy and Sell Stock v3

### ✅ Correct logic
- Single pass, O(n) time, O(1) space
- Correctly tracks `min_price` as the lowest price seen so far
- Correctly computes `profit` at each step and updates `max_profit`

### ❌ Bug: Initial `min_price` value
- You set `min_price = 1000`, but prices can be up to `10^4` (10,000).
- If all prices are above 1000 (e.g., `[5000, 6000]`), `min_price` will stay at 1000, giving a false profit.
