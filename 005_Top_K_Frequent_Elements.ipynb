## Problem: Top K Frequent Elements

Given an integer array `nums` and an integer `k`, return the `k` most frequent elements. You may return the answer in any order.

### Example 1:
```
Input: nums = [1,1,1,2,2,3], k = 2
Output: [1,2]
```

### Example 2:
```
Input: nums = [1], k = 1
Output: [1]
```

### Constraints:
- `1 <= nums.length <= 10^5`
- `-10^4 <= nums[i] <= 10^4`
- `k` is in the range `[1, the number of unique elements in the array]`
- It is guaranteed that the answer is unique.

### 1st Try

````python
def top_k_frequent_v1(nums, k):
    frequent = dict()
    for num in nums:
        if num not in frequent:
            frequent[num] = 1
        else:
            frequent[num] += 1
    return sorted(frequent, key=frequent.get, reverse=True)[:k]
````

````python
# Input: nums = [1,1,1,2,2,3], k = 2
# Output: [1,2]

nums = [1,1,1,2,2,3]
k = 2
top_k_frequent_v1(nums, k)
````

````python
nums = [1]
k = 1
top_k_frequent_v1(nums, k)
````

#### 1st Try is O(nlogn) since I used the sorted fucn. Although, the created dic is just O(n)

### 2nd Try

````python
def top_k_frequent_v2(nums, k):
    frequency = dict()
    for num in nums:
        if num not in frequency:
            frequency[num] = 1
        else:
            frequency[num] += 1

    buckets = [[] for _ in range(len(nums))]
    for key, value in frequency.items():
        buckets[value].append(key)

    ans = list()
    for i in range(len(buckets)-1, -1, -1):
        for num in buckets[i]:
            ans.append(num)
            k -= 1
        if k==0:
            return ans
````

````python
nums = [1,1,1,2,2,2,3]
k = 2
top_k_frequent_v2(nums, k)
````

## Best Practice

````python
from collections import Counter

def top_k_frequent(nums, k):
    # Step 1: Count frequencies
    freq = Counter(nums)

    # Step 2: Bucket sort by frequency
    buckets = [[] for _ in range(len(nums) + 1)]
    for num, count in freq.items():
        buckets[count].append(num)

    # Step 3: Collect top k from highest frequency
    result = []
    for i in range(len(buckets) - 1, 0, -1):
        for num in buckets[i]:
            result.append(num)
            if len(result) == k:
                return result
````
