## Problem: Group Anagrams

Given an array of strings `strs`, group the anagrams together. You can return the answer in any order.

An **anagram** is a word or phrase formed by rearranging the letters of a different word or phrase, using all the original letters exactly once.

### Example 1:
```
Input: strs = ["eat","tea","tan","ate","nat","bat"]
Output: [["bat"],["nat","tan"],["ate","eat","tea"]]
```

### Example 2:
```
Input: strs = [""]
Output: [[""]]
```

### Example 3:
```
Input: strs = ["a"]
Output: [["a"]]
```

### Constraints:
- `1 <= strs.length <= 10^4`
- `0 <= strs[i].length <= 100`
- `strs[i]` consists of lowercase English letters.

# 1st Try

````python
# Your implementation here
def group_anagrams(strs):
    ans = []
    anagram = []
    check = []
    for i in range(len(strs)):
        flag = False
        if strs[i] not in check:
            flag = True
            check.append(strs[i])
            anagram = [strs[i]]
            for j in range(i+1, len(strs)):
                if (strs[j] not in check and
                        sorted(strs[i])==sorted(strs[j])):
                    anagram.append(strs[j])
                    check.append(strs[j])
        if flag:
            ans.append(anagram)

    return ans
````

````python
# Input: strs = ["eat","tea","tan","ate","nat","bat"]
# Output: [["bat"],["nat","tan"],["ate","eat","tea"]]
strs = ["eat","tea","tan","ate","nat","bat"]
group_anagrams(strs)
````

````python
# Input: strs = [""]
# Output: [[""]]
strs = [""]
group_anagrams(strs)
````

````python
# Input: strs = ["a"]
# Output: [["a"]]

strs = ["a"]
group_anagrams(strs)
````

# 2nd Try:

#### the 1st Try works, however, I should use dictionaries

````python
# Your implementation here
def group_anagrams_v2(strs):
    ans = {}
    for word in strs:
        sort_word = "".join(sorted(word))
        if sort_word not in ans:
            ans [sort_word] = [word]
        else:
            ans[sort_word].append(word)

    return list(ans.values())
````

````python
strs = ["eat","tea","tan","ate","nat","bat"]
group_anagrams_v2(strs)
````

````python
strs = [""]
group_anagrams_v2(strs)
````

````python
strs = ["a"]
group_anagrams_v2(strs)
````

## Best Practice

````python
def group_anagrams_v3(strs):
    ans = {}
    for word in strs:
        ans.setdefault("".join(sorted(word)), []).append(word)
    return list(ans.values())
````
