# Question 1 — Two Sum (Easy)

## Problem
Given an array of integers `nums` and an integer `target`, return the **indices of the two numbers** such that they add up to `target`.

## Constraints / Notes
- Assume exactly **one solution exists** in the classic LeetCode version (but we’ll code it generally).
- You **cannot use the same element twice**.

## Example 1
```python
Input:
nums = [2, 7, 11, 15]
target = 9

Output:
[0, 1]

Explanation:
nums[0] + nums[1] = 2 + 7 = 9


## Mine Implementation

````python
from typing import List

def two_sum(nums: List[int], target: int) -> List[int]:
    for num in nums:
        find = target - num
        if find in nums:
            return [nums.index(num), nums.index(find)]
    return []
````

````python
two_sum([1, 3, 4, 5, 6], 9)  # return [1, 4]
````

````python
two_sum([7, 8, 9], 3)       # return []
````

# Modified Approach - Using Hash - Dictionaries

Complexity is O(n)

````python
from typing import List

def two_sum_hash_map(nums: List[int], target: int) -> List[int]:
    seen = {}
    for i, num in enumerate(nums):
        # print(seen)
        find = target - num
        if find in seen:
            return [seen[find], i]
        seen[num] = i # key, value
    return []
````

### Simple Checks

````python
two_sum_hash_map([1, 3, 4, 5, 6], 9)  # return [1, 4]
````

````python
two_sum_hash_map([7, 8, 9], 3)       # return []
````

````python
two_sum_hash_map([3, 2, 4], 6) # return [1, 2]
````

````python
two_sum_hash_map([3, 2, 4], 6) # return [1, 2]
````

````python
from typing import List

def two_sum(nums: List[int], target: int) -> List[int]:
    seen = {}

    for i, num in enumerate(nums):
        complement = target - num

        if complement in seen:
            return [seen[complement], i]

        seen[num] = i

    return []
````
