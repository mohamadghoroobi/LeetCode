## Question 2 — Valid Parentheses (Easy)
### Problem
Given a string containing just the characters '(', ')', '{', '}', '[', and ']', determine if the input string is valid.

A string is considered valid if:

Open brackets are closed by the same type of brackets.
Open brackets are closed in the correct order.
Constraints / Notes
The input string length is between 1 and 10^4.
The string may contain only these characters: '()[]{}.

## Example 1

### Input:
s = "()[]{}"

Output:
True

Explanation:
Every opening bracket is matched with the correct closing bracket in the correct order.


## Example 2

#### Input:
s = "([)]"

Output:
False

Explanation:
The brackets close in the wrong order: the `)` closes the `(` but the `]` closes the `[`, which is invalid here.

### Mine Solution

````python
# from typing import str

def is_valid(s: str) -> bool:
    storage = []
    for char in s:
        if char in "([{":
            storage.append(char)
        else:
            closing =  char
            opening = storage.pop()
            if opening == "(" and closing != ")":
                return False
            elif opening == "[" and closing != "]":
                return False
            elif opening == "{" and closing != "}":
                return False
            else:
                continue
    return True
````

````python
# check Example 1

s = "()[]{}"
is_valid(s)
````

````python
# check Example 2

s = "([)]"
is_valid(s)
````

## Best Practice

````python
# from typing import str

def is_valid(s: str) -> bool:
    stack = []
    pairs = {")":"(", "}":"{", "]":"[",}

    for char in s:
        if char in "([{":
            stack.append(char)


        # if not stack and stack.pop() != pairs[char]:
        #     return False
        else:
            if not stack:
                return False
            openings = stack.pop()
            if openings != pairs[char]:
                return False
    return len(stack) == 0
````
