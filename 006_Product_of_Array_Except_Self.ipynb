## Problem: Product of Array Except Self

Given an integer array `nums`, return an array `answer` such that `answer[i]` is equal to the product of all the elements of `nums` except `nums[i]`.

You must write an algorithm that runs in **O(n)** time and **without using the division operation**.

### Example 1:
```
Input: nums = [1,2,3,4]
Output: [24,12,8,6]
```

### Example 2:
```
Input: nums = [-1,1,0,-3,3]
Output: [0,0,9,0,0]
```

### Constraints:
- `2 <= nums.length <= 10^5`
- `-30 <= nums[i] <= 30`
- The product of any prefix or suffix of nums is guaranteed to fit in a 32-bit integer.

## 1st Try

````python
# Your implementation here
def product_except_self_v1(nums):
    ans = []
    for i in range(len(nums)):
        multiply = 1
        for j in range(len(nums)):
            if i != j:
                multiply *= nums[j]
        ans.append(multiply)
    return ans
````

````python
# Input: nums = [1,2,3,4]
# Output: [24,12,8,6]

nums = [1,2,3,4]
product_except_self_v1(nums)
````

````python
# Input: nums = [-1,1,0,-3,3]
# Output: [0,0,9,0,0]

nums = [-1,1,0,-3,3]
product_except_self_v1(nums)
````

## 2nd Try: getting hint of left and right calculation approach

````python
# Your implementation here
def product_except_self_v2(nums):
    left = [1] * len(nums)
    for i in range(1, len(nums)):
        left[i] = left[i-1] * nums[i-1]

    right = [1] * len(nums)
    for i in range(len(nums)-2, -1, -1):
        right[i] = right[i+1] * nums[i+1]

    answer = [0] * len(nums)
    for i in range(len(nums)):
        answer[i] = left[i] * right[i]

    return answer
````

````python
nums = [-1,1,0,-3,3]
product_except_self_v2(nums)
````

## Best Practice


````python
def product_except_self(nums):
    n = len(nums)
    answer = [1] * n

    # Left products directly in answer
    for i in range(1, n):
        answer[i] = answer[i-1] * nums[i-1]

    # Multiply by right products on the fly
    right = 1
    for i in range(n-1, -1, -1):
        answer[i] *= right
        right *= nums[i]

    return answer
````
