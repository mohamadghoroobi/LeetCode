## Problem: Longest Consecutive Sequence

Given an unsorted array of integers `nums`, return the length of the longest consecutive elements sequence.

You must write an algorithm that runs in **O(n)** time.

### Example 1:
```
Input: nums = [100,4,200,1,3,2]
Output: 4
Explanation: The longest consecutive elements sequence is [1, 2, 3, 4]. Therefore its length is 4.
```

### Example 2:
```
Input: nums = [0,3,7,2,5,8,4,6,0,1]
Output: 9
```

### Constraints:
- `0 <= nums.length <= 10^5`
- `-10^9 <= nums[i] <= 10^9`

## 1st Try

````python
def longest_consecutive_v1(nums):
    next_before_num = {}
    for num in nums:
        if num not in next_before_num and \
                (num - 1 in next_before_num or num + 1 in next_before_num):
            value = max(next_before_num[num - 1], next_before_num[num + 1])
            next_before_num[num] = value + 1

    return max(next_before_num.values())
````

````python
# Input: nums = [100,4,200,1,3,2]
# Output: 4
nums = [100,4,200,1,3,2]
longest_consecutive_v1(nums)
````

## Review: Longest Consecutive Sequence v1

### ❌ Bug: Only adds numbers that already have a neighbor
If a number is isolated (no `num-1` or `num+1` in the dict yet), you skip it entirely. But sequences can start from any number.

### 🔍 Test case that fails:
```
Input: nums = [1, 2, 3]
Your output: Error or incomplete dict
Expected: 3
```
- `num=1`: `2 not in dict` → skips, dict stays `{}`
- `num=2`: `1 not in dict` → skips
- `num=3`: skips
- Final dict is empty → `max({})` throws error

### ❌ Bug: Values don't track actual consecutive length
You take `max(value at num-1, value at num+1) + 1`, but those values don't represent the full length of their respective sequences — they represent something else. The logic for merging sequences is off.

### 💡 Hint:
1. Convert `nums` to a **set** for O(1) lookups
2. Only start counting from numbers that have **no left neighbor** (`num - 1 not in set`)
3. For each starting number, count up: `num + 1`, `num + 2`, ... until you hit a gap
4. Track the max length

This is O(n) because each element is visited at most twice.

## 2nd Try

````python
def longest_consecutive_v2(nums):
    num_set = set(nums)
    max_len = 0

    for num in num_set:
        if num - 1 not in num_set:      # start of a sequence?
            current_num = num
            current_len = 1
            while current_num + 1 in num_set:
                current_num += 1
                current_len += 1
            max_len = max(max_len, current_len)

    return max_len
````
