## Question 3 — Longest Substring Without Repeating Characters (Medium)
### Given a string s, find the length of the longest substring without repeating characters.

### A substring is a contiguous sequence of characters within a string.

### You must return the length of the longest such substring.

Example 1

``` python
Input:
s = “abcabcbb”

Output:

3

Explanation:

The answer is "abc", with length 3.

Example 2

### Better Solution with Sliding Windows

````python
def longest_substring(s):
    last_seen = dict()
    left = 0
    max_len = 0
    for right in range(len(s)):
        char = s[right]


        if char in last_seen:
            if last_seen[char] >= left:
                left = last_seen[char] + 1

        last_seen[char] = right

        window_length = right - left + 1
        max_len = max(max_len, window_length)

    return max_len
````
