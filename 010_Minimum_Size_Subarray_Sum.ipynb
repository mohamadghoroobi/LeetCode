## Problem: Minimum Size Subarray Sum

Given an array of positive integers `nums` and a positive integer `target`, return the minimal length of a subarray whose sum is greater than or equal to `target`. If there is no such subarray, return `0` instead.

### Example 1:
```
Input: target = 7, nums = [2,3,1,2,4,3]
Output: 2
Explanation: The subarray [4,3] has the minimal length under the problem constraint.
```

### Example 2:
```
Input: target = 4, nums = [1,4,4]
Output: 1
```

### Example 3:
```
Input: target = 11, nums = [1,1,1,1,1,1,1,1]
Output: 0
```

### Constraints:
- `1 <= target <= 10^9`
- `1 <= nums.length <= 10^5`
- `1 <= nums[i] <= 10^4`



### 1st Try

````python
# Your implementation here
def min_sub_array_len(nums, target):
    left = 0
    len_nums = len(nums)
    min_counter = len_nums + 1
    while (left < len_nums):
        sum = 0
        current_counter = 0
        right = left
        while (sum < target and right < len_nums):
            sum += nums[right]
            current_counter += 1
            right += 1
        if current_counter < min_counter and sum >= target:
            min_counter = current_counter
        left += 1
    return min_counter if min_counter != len_nums+1 else 0
````

````python
# Input: target = 7, nums = [2,3,1,2,4,3]
# Output: 2

target = 7
nums = [2, 3, 1, 2, 4, 3]

min_sub_array_len(nums, target)
````

````python
# Input: target = 4, nums = [1,4,4]
# Output: 1

target = 4
nums = [1, 4, 4]

min_sub_array_len(nums, target)
````

````python
# Input: target = 11, nums = [1,1,1,1,1,1,1,1]
# Output: 0

target = 11
nums = [1, 1, 1, 1, 1, 1, 1, 1]

min_sub_array_len(nums, target)
````

````python
# Input: target = 11, nums = [1,1,1,1,1,1,1,1]
# Output: 0

target = 15
nums = [1, 1, 1, 1, 1, 1, 1, 1,15]

min_sub_array_len(nums, target)
````

## 2nt Try: best practice with sliding window

````python
# Your implementation here
def min_sub_array_len_v2(nums, target):
    left = 0
    window_sum = 0
    min_len = float('inf')

    for right in range(len(nums)):
        window_sum += nums[right]          # expand

        while window_sum >= target:        # shrink
            min_len = min(min_len, right - left + 1)
            window_sum -= nums[left]
            left += 1

    return min_len if min_len != float('inf') else 0
````

````python
# Input: target = 7, nums = [2,3,1,2,4,3]
# Output: 2

target = 7
nums = [2, 3, 1, 2, 4, 3]

min_sub_array_len(nums, target)
````
