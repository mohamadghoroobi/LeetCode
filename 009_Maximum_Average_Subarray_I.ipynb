## Problem: Maximum Average Subarray I

You are given an integer array `nums` consisting of `n` elements, and an integer `k`.

Find a contiguous subarray whose length is equal to `k` that has the maximum average value and return this value. Any answer with a calculation error less than `10^-5` will be accepted.

### Example 1:
```
Input: nums = [1,12,-5,-6,50,3], k = 4
Output: 12.75000
Explanation: Maximum average is (12 - 5 - 6 + 50) / 4 = 51 / 4 = 12.75
```

### Example 2:
```
Input: nums = [5], k = 1
Output: 5.00000
```

### Constraints:
- `n == nums.length`
- `1 <= k <= n <= 10^5`
- `-10^4 <= nums[i] <= 10^4`



## 1st Try

````python
# Your implementation here
def find_max_average(nums, k):
    # if k == 1:
    #     return nums[0]
    ans = list()
    max = 0
    for i in range(len(nums) - k + 1):
        sum = 0
        for j in range(i, i + k):
            sum += nums[j]
        if sum > max:
            max = sum
    return max / k
````

````python
# Input: nums = [1,12,-5,-6,50,3], k = 4
# Output: 12.75000
nums = [1, 12, -5, -6, 50, 3]
k = 4
find_max_average(nums, k)
````

````python
# Input: nums = [5], k = 1
# Output: 5.00000
nums = [5]
k = 1
find_max_average(nums, k)
````

## 2nt Try: Sliding Window

````python
# Your implementation here
def find_max_average(nums, k):
    sum_old = 0
    for i in range(k):
        sum_old += nums[i]
    len_nums = len(nums)
    left = 0
    right = k
    max = sum_old
    while right < len_nums:
        sum_old = sum_old + nums[right] - nums[left]
        if sum_old > max:
            max = sum_old
        left += 1
        right += 1
    return max / k
````

````python
# Input: nums = [1,12,-5,-6,50,3], k = 4
# Output: 12.75000
nums = [1, 12, -5, -6, 50, 3]
k = 4
find_max_average(nums, k)
````

## 3rd Try: Better

````python
# Your implementation here
def find_max_average(nums, k):
    window_sum = sum(nums[:k])
    max_sum = window_sum
    for i in range(k, len(nums)):
        window_sum += nums[i] - nums[i - k]
        if window_sum > max_sum:
            max_sum = window_sum
    return max_sum / k
````

````python
# Input: nums = [1,12,-5,-6,50,3], k = 4
# Output: 12.75000
nums = [1, 12, -5, -6, 50, 3]
k = 4
find_max_average(nums, k)
````
