## Hint: Prefix Sum + Hash Map

### What is a Prefix Sum?

A prefix sum is the sum of all elements from the start up to a given index.

```
nums =   [1, 2, 3]
prefix = [1, 3, 6]
           ↑  ↑  ↑
           sum of [1]
              sum of [1,2]
                 sum of [1,2,3]
```

---

### How it helps find subarray sums:

The sum of subarray from index `i+1` to `j` is: **`prefix[j] - prefix[i]`**

```
nums =  [1, 2, 3]
prefix = [1, 3, 6]

Subarray [2, 3] = prefix[2] - prefix[0] = 6 - 1 = 5 ✅
```

---

### Applying to the problem:

You want `sum(nums[i+1 : j+1]) == k`

That means: **`prefix[j] - prefix[i] == k`**

Rearrange: **`prefix[i] == prefix[j] - k`**

---

### The trick:

As you iterate through the array, keep a running `current_sum` (prefix up to current index). At each step, ask:

> "Have I seen a prefix sum equal to `current_sum - k` before? How many times?"

Store prefix sums and their counts in a hash map.

---

### Walkthrough for `nums = [1, 2, 3], k = 3`:

| i | num | current_sum | current_sum - k | Seen in map? | Count added | Map |
|---|-----|-------------|-----------------|--------------|-------------|-----|
| 0 | 1 | 1 | -2 | No | 0 | {0:1, 1:1} |
| 1 | 2 | 3 | 0 | Yes (0 → 1 time) | 1 | {0:1, 1:1, 3:1} |
| 2 | 3 | 6 | 3 | Yes (3 → 1 time) | 1 | {0:1, 1:1, 3:1, 6:1} |

Total count = 2 (subarrays [1,2] and [3])

---

### 💡 You need to figure out:
1. What initial value goes in the hash map? (Hint: empty subarray sum = 0)
2. When do you update the map?
3. How to handle the running sum?

````python
def subarray_sum(nums, k):
    map = {0:1}
    count = 0
    current_sum = 0

    for i in range(len(nums)):
        current_sum += nums[i]
        need = current_sum - k
        if need in map:
            count += 1
        map.update({current_sum:1})
    return count
````

````python
nums =   [1, 2, 3]
k = 2
subarray_sum(nums, k)
````

````python
nums = [3, 4, -2, 1, 2]
k = 3
subarray_sum(nums, k)
````

## Try with indices

````python
def subarray_sum_indices(nums, k):
    prefix_map = {0: [-1]}  # prefix sum 0 at index -1
    current_sum = 0
    result = []

    for i in range(len(nums)):
        current_sum += nums[i]
        need = current_sum - k

        if need in prefix_map:
            for start_idx in prefix_map[need]:
                result.append(nums[start_idx + 1 : i + 1])

        if current_sum not in prefix_map:
            prefix_map[current_sum] = []
        prefix_map[current_sum].append(i)

    return result
````

````python
nums = [3, 4, -2, 1, 2]
k = 3
subarray_sum_indices(nums, k)
````
